Here are request params constants to use later. I also have separate bands of different trading volumes to try to get an even spread of markets with different volumes to show differences between high, medium and low markets.

```python
BASE_URL = "https://gamma-api.polymarket.com"
LIMIT = 100
CLOSED = "true"
INCLUDE_TAG = "true"

PERIOD_START = date(2025, 1, 1)
PERIOD_END = date(2026, 10, 31)
NUM_WINDOWS = 6
WEEK_DAYS = 7
SEED = 42

VOLUME_BANDS = [
    (100, 1_000),
    (1_000, 10_000),
    (10_000, 100_000),
    (100_000, 1_000_000),
    (1_000_000, None),
]
BAND_LABELS = ["$100–1k", "$1k–10k", "$10k–100k", "$100k–1M", "$1M+"]

OFFSETS = [0, 100, 200, 300, 400, 500]
```

Without getting random dates the API gives pretty bad small markets from the past day that the request was made so I am generating a bunch of random date ranges to get better data

```python
random.seed(SEED)
segment_days = (PERIOD_END - PERIOD_START).days // NUM_WINDOWS

# One random 7-day window inside each segment
WEEK_STARTS, WEEK_ENDS = [], []
for i in range(NUM_WINDOWS):
    segment_start = PERIOD_START + timedelta(days=i * segment_days)
    week_start = segment_start + timedelta(days=random.randint(0, segment_days - WEEK_DAYS))
    WEEK_STARTS.append(week_start)
    WEEK_ENDS.append(week_start + timedelta(days=WEEK_DAYS))

# One random whole month inside each segment
MONTH_STARTS, MONTH_ENDS = [], []
for i in range(NUM_WINDOWS):
    segment_start = PERIOD_START + timedelta(days=i * segment_days)
    day_in_segment = segment_start + timedelta(days=random.randint(0, segment_days - 1))
    month_start = day_in_segment.replace(day=1)
    next_month = (month_start + timedelta(days=32)).replace(day=1)
    MONTH_STARTS.append(month_start)
    MONTH_ENDS.append(next_month)

print("Weeks: ", [f"{s} → {e}" for s, e in zip(WEEK_STARTS, WEEK_ENDS)])
print("Months:", [f"{s:%Y-%m}" for s in MONTH_STARTS])
```

# Combining the Bands

```python
markets_raw = pd.concat([band1, band2, band3, band4, band5], ignore_index=True)
markets_raw = markets_raw.drop_duplicates('id')
print('Combined shape:', markets_raw.shape)
markets_raw['volume_band'].value_counts()
```

```python
markets = markets_raw.reindex(columns=['id', 'question', 'startDate', 'endDate', 'volumeNum', 'volume_band',
                       'outcomes', 'outcomePrices', 'negRisk', 'tags']).copy()  # reindex: a missing field becomes an empty column instead of an error
markets = markets.rename(columns={'id': 'market_id', 'startDate': 'start_date', 'endDate': 'end_date',
                                  'volumeNum': 'volume_usd', 'negRisk': 'neg_risk'})
markets
```

# Cleanup

```python
markets['outcomes'] = markets['outcomes'].apply(lambda s: json.loads(s) if isinstance(s, str) else [])
markets['outcomePrices'] = markets['outcomePrices'].apply(lambda s: [float(p) for p in json.loads(s)] if isinstance(s, str) else [])

markets['resolved_outcome'] = markets.apply(
    lambda row: row['outcomes'][row['outcomePrices'].index(max(row['outcomePrices']))]
    if row['outcomes'] and len(row['outcomes']) == len(row['outcomePrices']) else None, axis=1)
markets[['question', 'outcomes', 'outcomePrices', 'resolved_outcome']].head(5)
```

```python
markets['tags'] = markets['tags'].apply(lambda ts: [t['label'] for t in ts] if isinstance(ts, list) else [])
markets['tag_count'] = markets['tags'].apply(len)

market_tags = markets[['market_id', 'tags']].explode('tags').rename(columns={'tags': 'tag'}).dropna()
print('market_tags shape:', market_tags.shape)
market_tags['tag'].value_counts().head(20)
```

```python
print('Missing values before cleaning:')
print(markets[['start_date', 'end_date', 'volume_usd']].isna().sum())

markets['start_date'] = pd.to_datetime(markets['start_date'], errors='coerce', utc=True).dt.tz_localize(None)
markets['end_date'] = pd.to_datetime(markets['end_date'], errors='coerce', utc=True).dt.tz_localize(None)
markets = markets.dropna(subset=['start_date', 'end_date', 'volume_usd'])

markets['duration_days'] = (markets['end_date'] - markets['start_date']).dt.days
markets = markets[markets['duration_days'] >= 0]
markets['start_month'] = markets['start_date'].dt.to_period('M').dt.to_timestamp()
markets['start_month_name'] = markets['start_date'].dt.month_name()
markets['tags'] = markets['tags'].apply(', '.join)   # readable text version for Power BI

print('Final shape:', markets.shape)
markets.head()
```
